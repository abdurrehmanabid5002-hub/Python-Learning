```python
data = {
    "Over": np.arange(1, 21),
    "CSK_Runs": [
        7, 8, 5, 10, 6,
        9, 4, 12, 7, 8,
        11, 6, 9, 10, 5,
        8, 12, 7, 10, 11
    ],
    "MI_Runs": [
        5, 7, 8, 6, 9,
        4, 10, 7, 8, 11,
        5, 9, 6, 12, 7,
        10, 5, 11, 8, 9
    ]
}

df = pd.DataFrame(data)

print("IPL Match Data:")
print(df)

csk_total = np.sum(df["CSK_Runs"])
mi_total = np.sum(df["MI_Runs"])

csk_avg = np.mean(df["CSK_Runs"])
mi_avg = np.mean(df["MI_Runs"])

df["CSK_Cumulative"] = np.cumsum(df["CSK_Runs"])
df["MI_Cumulative"] = np.cumsum(df["MI_Runs"])

print("\nMatch Summary")
print("-" * 30)
print("CSK Total Runs:", csk_total)
print("MI Total Runs:", mi_total)
print("CSK Average Runs/Over:", round(csk_avg, 2))
print("MI Average Runs/Over:", round(mi_avg, 2))

if csk_total > mi_total:
    print("Winner: CSK")
elif mi_total > csk_total:
    print("Winner: MI")
else:
    print("Match Draw")

plt.figure(figsize=(10, 5))

plt.plot(
    df["Over"],
    df["CSK_Runs"],
    marker="o",
    label="CSK"
)

plt.plot(
    df["Over"],
    df["MI_Runs"],
    marker="o",
    label="MI"
)

plt.title("IPL Match - Runs Per Over")
plt.xlabel("Over")
plt.ylabel("Runs")
plt.xticks(df["Over"])
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 5))

plt.plot(
    df["Over"],
    df["CSK_Cumulative"],
    marker="o",
    label="CSK"
)

plt.plot(
    df["Over"],
    df["MI_Cumulative"],
    marker="o",
    label="MI"
)

plt.title("IPL Match - Cumulative Runs")
plt.xlabel("Over")
plt.ylabel("Total Runs")
plt.xticks(df["Over"])
plt.legend()
plt.grid(True)
plt.show()

teams = ["CSK", "MI"]
total_runs = [csk_total, mi_total]

plt.figure(figsize=(7, 5))

plt.bar(teams, total_runs)

plt.title("IPL Match - Total Runs")
plt.xlabel("Team")
plt.ylabel("Total Runs")

plt.show()
```
