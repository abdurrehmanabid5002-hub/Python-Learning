```python
df.isnull().sum()
```

```python
DUP=df.duplicated().sum()

print(DUP)
```

```python
run=df["runs"].cumsum().astype("Float64")
```

```python
plt.plot(run)
# plt.hist(cumsum_runs,bins=[0,20,30,40,50,60,70,80,100])
plt.title("PSL Data Analysis Using NumPy")
plt.xlabel("Players ID")
plt.ylabel("Runs")
plt.grid(False)
plt.show
```

```python
df["avg_bat"].unique()
df["avg_bat"].value_counts()
```

```python
Avgbat_graph=df["avg_bat"].cumsum().astype("Float64")
plt.plot(Avgbat_graph)
plt.title("PSL Data Analysis Using NumPy")
plt.xlabel("Players ID")
plt.ylabel("Runs")
plt.grid(False)
plt.show
```

```python
Strikerate_graph=df["strike_rate"].cumsum().astype("Float64")
plt.plot(Strikerate_graph)
plt.title("PSL Data Analysis Using NumPy")
plt.xlabel("Players ID")
plt.ylabel("Runs")
plt.grid(False)
plt.show
```

```python
Strikerate_graph=df["strike_rate"].cumsum().astype("Float64")
plt.plot(Strikerate_graph)
plt.plot(Avgbat_graph)
plt.plot(run)
plt.title("PSL Data Analysis Using NumPy")
plt.xlabel("Players ID")
plt.ylabel("Runs")
plt.grid(False)
plt.legend()
plt.show
```

```python
fig ,axes=plt.subplots(1,3,figsize=(14,4))
axes[0].plot(run,color="blue")
axes[0].set_title("Graph 1")

axes[1].plot(Avgbat_graph,color="red")
axes[1].set_title("Graph 2")

axes[2].plot(Strikerate_graph,color="green")
axes[2].set_title("Graph 3")
```
