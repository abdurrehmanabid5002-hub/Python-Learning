# Numpy Functions


```python
zeros_array=np.zeros((4,4))
print("zeros array :", zeros_array)

ones_array=np.ones((3,3))
print("\nOnes array :", ones_array)

full_array=np.full((3,3),7)
print("\nFull Array :",full_array)

random_array=np.random.random((3,3))
print("\nRandom array",random_array)

sequence_array=np.arange(0,11,2)
print("\nSquence Array:",sequence_array)
```

# Vector, Matrices and Tesnor

```python
vector=np.array([1,2,3])
print("Vector= ",vector)

matrices=np.array([[1,2,3],[4,5,6]])
print("Matrices= ",matrices)

tensor=np.array([[[1,2,3],[1,2,3]],
                 [[1,2,3],[1,2,3]]])
print("Tensor= ",tensor)
print(tensor.shape)
```
