```python
import numpy as np
arr=np.arange((12))
print("array ", arr)


reshaped=arr.reshape((3,4))
print("Reshape array",reshaped)

flat=reshaped.flatten()
print("Flatend ",flat)

transpose=reshaped.T
print ("The transpose of reshapes matrice is ", transpose)
```

```python
vector1=np.array([1,2,3,4,5])
vector2=np.array([6,7,8,9,10])
print("Vector addition\n",vector1+vector2)
print("Vector Multiplication\n",vector1*vector2)
print("Dot Product of two vectors\n",np.dot(vector1,vector1))

angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print(angle)

print("concatition of array\n",np.concatenate((vector1,vector2)))
```

## Sin Graph 

```python


x = np.arange(0, 3*np.pi, 0.1)
y = np.sin(x)

plt.figure(figsize=(8,4))
plt.plot(x, y)

plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True)

plt.show()
```

## Dark Logo 


```python
try:
    simple_logo=np.load('numpy-logo.npy')
    plt.figure(figsize=(8,4))
    plt.subplot(121)
    plt.imshow(simple_logo)
    plt.title("NumPy Logo")
    plt.grid(False)
    plt.show()
    
    Dark_Logo=1-simple_logo
    plt.figure(figsize=(8,4))
    plt.subplot(121)
    plt.imshow(Dark_Logo)
    plt.title("NumPy Dark Logo")
    plt.grid(False)
    plt.show()

    

except FileNotFoundError:
    print("file not found error ")
```
